```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
mydf = pd.read_csv("robot_battery_data.csv")

mydf.info()
```

```python
mydf.head()
```

```python
X_axis = mydf[['payload_kg', 'speed_mps', 'motor_temp_c', 'light_lux', 'motor_current_a']]
```

```python
X_axis.head()
```

```python
y_axis = mydf[['battery_drain']]
```

```python
y_axis.head()
```

```python
my_model = LinearRegression().fit(X_axis, y_axis)
```

```python
my_model
```

```python
print(my_model.coef_)
```

```python
print(my_model.intercept_)
```

```python
payload_grid = np.linspace(X_axis["payload_kg"].min(), X_axis["payload_kg"].max(), 30)
speed_grid = np.linspace(X_axis["speed_mps"].min(), X_axis["speed_mps"].max(), 30)
motor_temp_c_grid = np.linspace(X_axis["motor_temp_c"].min(), X_axis["motor_temp_c"].max(), 30)
light_lux_grid = np.linspace(X_axis["light_lux"].min(), X_axis["light_lux"].max(), 30)
motor_current_a_grid = np.linspace(X_axis["motor_current_a"].min(), X_axis["motor_current_a"].max(), 30)
#speed_grid = np.linspace(X_axis["speed_mps"].min(), X_axis["speed_mps"].max(), 30)
```

```python
P, S, M, L, A= np.meshgrid(payload_grid, speed_grid, motor_temp_c_grid, light_lux_grid, motor_current_a_grid)
```

```python
grid_data = pd.DataFrame({
    "payload_kg": P.ravel(),
    "speed_mps": S.ravel(),
    "motor_temp_c": M.ravel(),
    "light_lux": L.ravel(),
    "motor_current_a": A.ravel(),
    
})
```

```python
Z = my_model.predict(grid_data).reshape(P.shape)
```
